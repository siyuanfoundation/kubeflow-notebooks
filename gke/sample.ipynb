```python
# a sample code to test pandas data analysis
import numpy as np
import pandas as pd

# ==========================================
# 1. LOADING & INSPECTING DATA
# ==========================================

# Create a sample DataFrame mimicking retail sales data
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107],
    "Department": [
        "Sales",
        "Marketing",
        "Sales",
        "HR",
        "Marketing",
        "Sales",
        "HR",
    ],
    "Salary": [70000, 62000, 81000, np.nan, 58000, 75000, 52000],
    "Performance_Score": [8.5, 7.2, 9.1, 6.5, np.nan, 8.8, 7.0],
    "Hire_Date": [
        "2021-01-15",
        "2022-03-22",
        "2020-07-11",
        "2023-01-10",
        "2021-11-05",
        "2019-04-23",
        "2023-05-19",
    ],
}

# Load into a pandas DataFrame (normally you would use pd.read_csv('file.csv'))
df = pd.DataFrame(data)

print("--- FIRST 3 ROWS ---")
print(df.head(3))  # View the top 3 rows

print("\n--- DATASET SUMMARY ---")
print(df.info())  # Check data types and missing values

print("\n--- DESCRIPTIVE STATISTICS ---")
print(df.describe())  # Get statistical metrics for numeric columns

# ==========================================
# 2. DATA CLEANING & TRANSFORMATION
# ==========================================

# Convert date column from string to datetime object
df["Hire_Date"] = pd.to_datetime(df["Hire_Date"])

# Handle missing data (Imputation)
# Fill missing salaries with the column median
salary_median = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(salary_median)

# Fill missing performance scores with the column mean
score_mean = df["Performance_Score"].mean()
df["Performance_Score"] = df["Performance_Score"].fillna(score_mean)

# Create a new feature (column) based on a condition
df["Is_Senior"] = df["Hire_Date"].dt.year < 2022

print("\n--- CLEANED DATAFRAME ---")
print(df)

# ==========================================
# 3. FILTERING & SORTING
# ==========================================

# Filter: High earners in the Sales department
high_sales = df[(df["Department"] == "Sales") & (df["Salary"] > 72000)]

# Sort the filtered results by performance score descending
high_sales_sorted = high_sales.sort_values(
    by="Performance_Score", ascending=False
)

print("\n--- TOP SALES PERFORMERS ---")
print(high_sales_sorted)

# ==========================================
# 4. AGGREGATION & GROUPING
# ==========================================

# Group by department and calculate aggregate metrics
dept_summary = (
    df.groupby("Department")
    .agg(
        Average_Salary=("Salary", "mean"),
        Max_Performance=("Performance_Score", "max"),
        Employee_Count=("Employee_ID", "count"),
    )
    .reset_index()
)

print("\n--- DEPARTMENTAL SUMMARY ---")
print(dept_summary)

# ==========================================
# 5. EXPORTING DATA
# ==========================================

# Save results to a CSV file (index=False prevents writing row numbers)
dept_summary.to_csv("department_analytics_report.csv", index=False)
print("\n[SUCCESS] Summary report saved to CSV.")
```

```python
# ==========================================
# 6. DATA VISUALIZATION
# ==========================================
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for clean visual aesthetics
sns.set_theme(style="whitegrid")

# Create a figure with two subplots side-by-side
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Average Salary by Department (Bar Plot)
sns.barplot(
    ax=axes[0],
    data=df,
    x="Department",
    y="Salary",
    errorbar=None,
    palette="viridis"
)
axes[0].set_title("Average Salary by Department", fontsize=14, pad=15)
axes[0].set_ylabel("Salary ($)", fontsize=12)
axes[0].set_xlabel("Department", fontsize=12)

# Plot 2: Salary vs Performance Score (Scatter Plot with regression line)
sns.regplot(
    ax=axes[1],
    data=df,
    x="Performance_Score",
    y="Salary",
    scatter_kws={"s": 80, "alpha": 0.8},
    line_kws={"color": "#e74c3c", "linewidth": 2}
)
axes[1].set_title("Salary vs. Performance Score", fontsize=14, pad=15)
axes[1].set_xlabel("Performance Score", fontsize=12)
axes[1].set_ylabel("Salary ($)", fontsize=12)

plt.tight_layout()
plt.show()
```
